```python
# Read the existing candidate pairs table
CandidatePairsDf = spark.read.table("semantic.ProductCandidatePairs")

# Export candidate pairs to the AI exchange folder
CandidatePairsDf.coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", True) \
    .option("escape", '"') \
    .option("quote", '"') \
    .csv("Files/AI_Exchange/Input/CandidatePairs_Export")
```
